### Exercise 5

For k = 1, 3, 7, 13, 18, what is this set in \\(F_{19}\\)?

{k⋅0, k⋅1, k⋅2, k⋅3, ... k⋅18}

Do you notice anything about these sets?

```python
# Exercise 5

prime = 19
for k in [1, 3, 7, 13, 18]:
    values = [(k * n) % prime for n in range(prime)]
    print(k, sorted(values))
```
